# Exercises

It's time to practice on your own. Write queries that return:

1. Orders above $300 in descending order by the sale amount from highest to lowest.

```python
order_total = func.sum(OrderItem.unit_price * OrderItem.quantity).label(None)
q = (select(Order, order_total)
     .join(OrderItem)
     .group_by(Order)
     .having(order_total > 300)
     .order_by(order_total.desc())
     )

session.execute(q).all()
```

2. Orders that include one or more ZX81 computers.

```python
num_products = func.count(Product.id).label(None)
q = (select(Order, num_products, Product.name)
     .join(Order.order_items)
     .join(OrderItem.product)
     .where(Product.name.ilike('%ZX%81%'))
     .group_by(Order)
     .having(num_products >= 1)
     )

session.execute(q).all()
```
