# Exercises

Now is your chance to practice some one-to-many relationship queries. Before starting, run the `import_products.py` script to restore any products or manufacturers you may have deleted. Then open a Python session and write queries that return:

1. The list of products made by IBM and Texas Instruments

```python
from db import Session
from models import Product, Manufacturer
session = Session()
```

```python
from sqlalchemy import select, or_
q = (select(
    Product).join(Manufacturer.products)
    .where(or_(Manufacturer.name.ilike("%IBM%"), Manufacturer.name.ilike("%Texas%Inst%")))
    .distinct()
)
session.execute(q).all()
```

2. Manufacturers that operate in Brazil

```python
from sqlalchemy import select, or_
q = (select(
    Manufacturer).join(Product.manufacturer)
    .where(Product.country.ilike("%Brazil%"))
    .distinct()
)
session.execute(q).all()
```

3. Products that have a manufacturer that has the word "Research" in their name.

```python
from sqlalchemy import select
q = (select(
    Product).join(Product.manufacturer)
    .where(Manufacturer.name.ilike("%Research%"))
)
session.execute(q).all()
```

4. Manufacturers that made products based on the Z80 CPU or any of its clones

```python
from sqlalchemy import select
q = (select(
    Manufacturer).join(Product.manufacturer)
    .where(Product.cpu.ilike("%Z80%"))
    .distinct()
)
session.execute(q).all()
```

5. Manufacturers that made products that are NOT based on the 6502 CPU or any of its clones.

```python
from sqlalchemy import select
q = (select(
    Manufacturer).join(Product.manufacturer)
    .where(~Product.cpu.ilike("%6502%"))
    .distinct()
)
session.execute(q).all()
```

6. Manufacturers and the year they went to market with their first product, sorted by the year.

```python
from sqlalchemy import func

q = (select(
    Manufacturer, func.min(Product.year)).join(Product.manufacturer)
).group_by(Manufacturer).order_by(func.min(Product.year))
session.execute(q).all()
```

7. Manufacturers that have 3 to 5 products in the catalog

```python
from sqlalchemy import func


q = (select(
    Manufacturer, func.count(Product.id).label('num_products')).join(Product.manufacturer)
).group_by(Manufacturer).having(func.count(Product.id).between(3,5))
session.execute(q).all()
#print(q)
```

8. Manufacturers that operated for more than 5 years.

```python
q = (
    select(
        Manufacturer,
        func.min(Product.year).label('start_year'),
        func.max(Product.year).label('end_year')
    )
    .join(Product.manufacturer)
    .group_by(Manufacturer)
    .having((func.max(Product.year) - func.min(Product.year)) > 5)
)

session.execute(q).all()
```

```python
from sqlalchemy import select, func, distinct

stmt = (
    select(Manufacturer, func.count(distinct(Product.year)))
    .join(Manufacturer.products)
    .group_by(Manufacturer.id)
    .having(func.count(distinct(Product.year)) > 5)
)

session.execute(stmt).all()
```
